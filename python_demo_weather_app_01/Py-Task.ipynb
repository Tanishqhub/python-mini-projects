```python
def auth(user):
    if user.isdigit():
        if len(user) == 1:
            user = int(user)
            if user>0 and user <5:
                print("Weather code authenticated...all ok!")
            else:
                print('Only code between 1 to 4 are allowed')
        else:
            print("Only one code is allowed")
    else:
        print("Please enter only numeric codes")
```

```python
print('='*100)
print("Welcome to Weather Guide")
print('='*100)
print("\nChoose your weather condition from following weather codes\n1. Rainy\n2. Sunny\n3. Windy\n4. Other weather condition")
weather = input("\nEnter your weather code: ")
auth(weather)

if weather == 1:
    rain = True
elif weather == 2:
    sunny = True
elif weather == 3:
    windy = True

if rain:
    if umbrella:
        print("Use Umbrella")
    elif raincoat:
        print("Use raincoat")
    else:
        print("Bahar nahi jana")
elif windy:
    if jacket:
        print("Wear Jacket")
    else:
        print("Stay Indoor, Stay warm")
elif sunny:
    if cap:
        print("Wear Cap")
    else:
        print("Wear goggles")
else:
    print("It's a pleasent weather, enjoy it")
```
