#### 3. Loop over a string

```python
for i in 'tanishq': # this line decides
    print(i) # how many times will this line execute

# the iterator holds the value of each item in the iterable and updates in each iteration
# the iterator can either use those values, or execute the given code iteration {len(iterable)} number of time
```

#### Continue statement - skips the iteration

```python
for i in 'tanishq':
    if i == 's':
        print(i)
    else:
        continue
```

#### Break statement - kills the loop

```python
for i in 'tanishq':
    if i != 's':
        print(i)
    else:
        break
```

## CH-2: String concatenation

#### Using + operation

```python
first_name = 'tanishq'
last_name = 'khandar'
```

```python
full_name = first_name + last_name # '+' is defined for cancatenating two string in python
full_name
```

```python
first_name = 'tanishq'
last_name = 'khandar'
full_name = first_name + ' ' + last_name # '+' is defined for cancatenating two string in python
full_name
```

#### Using print function

```python
first_name = 'tanishq'
last_name = 'khandar'
print('My name is',first_name,last_name) # this is not concatenation
```

#### Using .join

```python
first_name = 'tanishq'
last_name = 'khandar'
full_name = "_".join([first_name,last_name])
full_name
```

#### Using % operator

```python
first_name = input("What is your name: ")
"My name is %s" % first_name
```

#### Using .format

```python
"My name is {} {}".format(first_name,last_name)
```

#### Using f-strings

```python
f"My name is {first_name} {last_name}"
```
