### Running app multiple times - While loop method

```python
valid_users = ['tanishq', 'chirag', 'amol', 'bhavani', 'caren', 'ayushman', 'roshni']
counter = 0
app = True

while app:
    counter +=1
    print(f"App run count {counter}")
    user_name = input("Enter your name: ")
    
    while user_name.lower() not in valid_users:
        user_name = input("Renter your name: ")
    else:
        n1 = input("Enter n1: ")
        n2 = input("Enter n2: ")
    
        while n1.isdigit() == 0 or n2.isdigit() == 0:
            print("Kindly use brain and enter integers")
            n1 = input("Renter n1: ")
            n2 = input("Renter n2: ")
        else:
            n1 = int(n1)
            n2 = int(n2)
            print(f"Your output is {n1 + n2}")
    print()
    inp = input("Do you want to quit (y or leave as it is): ")
    if inp.lower() == 'y':
        print("Thanks for wasting your precious time")
        app = False
```
