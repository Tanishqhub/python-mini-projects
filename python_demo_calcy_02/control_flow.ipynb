```python
# context

# marks below 50 = fail
# below 65 = d
# below 75 = c
# below 85 = b
# below 95 = a
# above 95 = A+
```

```python
# Welcome part
print('Welcome to dumb grader:\n')
student = input('Enter your name: ')
print(f"\nwelcome {student}\n")
marks = int(input('Enter your marks: \n'))
print(f'Congratulations on scoring {marks} marks!!')

# Grading part
if marks > 95:
    print(f'Your grade is A+\nParty Bro')
elif marks > 85:
    print(f'Your grade is A\nMumma is proud of you')
elif marks > 75:
    print(f'Your grade is B\nGood lad, do better next time')
elif marks > 65:
    print(f'Your grade is C\nDont use insta')
elif marks > 50:
    print(f'Your grade is D\nYou are disgracefull, bring your parents')
else:
    print('You failed, You dont deserve education')
```
