### Example 2

```python
def power(number, power = 2):
    return number ** power
```

```python
power(2,3)
```

```python
power(3,2)
```

```python
power(1,2,3)
```

#### keyword arguments

```python
power(3,2)
```

```python
power(power=3,number=2)
```

```python
power(2 ,number = 3)
```

```python
power(number=2, 3)
```

```python
power(power=2, 3)
```

```python
power(number=2, power=4)
```

```python
op = power(5) # default argument
```

```python
print(f"My output is {op}")
```

```python
op + 10
```

```python
add = lambda n1,n2 : n1 + n2
```

```python
add(2,4)
```

```python
sqr = lambda x : x**2
```

```python
sqr(5)
```

```python
iterable = [1,2,3]
```

```python
list(map(sqr,iterable))
```

```python
tuple(map(lambda x: x**2,[1,2,3]))
```

```python
chars = ['t','a','n','i','s','h','q']
```

```python
from functools import reduce
```

```python
reduce(lambda x,y:x+y, chars)
```

```python
nums = [1,2,3,4,5]
```

```python
reduce(lambda x,y: x+y, nums)
```

```python
reduce(lambda x,y: x*y, nums)
```

```python
list(filter(lambda x:x > 3, nums))
```
